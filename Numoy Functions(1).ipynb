```python
import numpy as np
a = np.random.randint(1,9, size = (3,3))
print(a)
b = np.ones([3,3])
print(b)
add_ = np.add(a,b)
sub_ = np.subtract(a,b)
print(add_)
print(sub_)
```

```python
import numpy as np

a = np.array([1, 1, 2, 3, 3, 3, 4, 5, 6, 7, 7, 7, 8, 8, 9])

values, counts = np.unique(a, return_counts=True)
values = values[counts > 1] # 1 > 1 | 2 > 1 != 1 | 3 > 1 |
count = counts[counts > 1] # 1 > 1 = 2 | 3 > 1 = 3 |
print("Values :", values)
print("Count :", count)
```
